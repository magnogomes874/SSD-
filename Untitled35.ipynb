1. Preparar ambiente e carregar dados
Execute as células em ordem. Faça upload do arquivo ANAC_DADOS_AGREGADOS_ML.zip quando solicitado.

```python

import pandas as pd
import zipfile
import io
from pathlib import Path
from google.colab import files

print("Selecione os 5 arquivos ZIP:")
print("ANAC_2019, ANAC_2020, ANAC_2022, ANAC_2023 e ANAC_2025")

enviados = files.upload()
bases = []

for nome, conteudo in enviados.items():
    if nome.lower().endswith(".zip"):
        with zipfile.ZipFile(io.BytesIO(conteudo)) as z:
            arquivos_csv = [
                a for a in z.namelist()
                if a.lower().endswith(".csv")
            ]

            if not arquivos_csv:
                print(f"Nenhum CSV encontrado em {nome}")
                continue

            for csv in arquivos_csv:
                df = pd.read_csv(z.open(csv))
                bases.append(df)
                print(f"Importado: {nome} — {len(df)} registros")

    elif nome.lower().endswith(".csv"):
        df = pd.read_csv(io.BytesIO(conteudo))
        bases.append(df)
        print(f"Importado: {nome} — {len(df)} registros")

if not bases:
    raise ValueError("Nenhuma base válida foi importada.")

full = pd.concat(bases, ignore_index=True)

# Salvar a base consolidada
full.to_csv(
    "/content/anac_agregada_ml.csv",
    index=False
)

# Preparar dados de novembro
nov = full[full["mes"] == 11].copy()

print("\nIMPORTAÇÃO CONCLUÍDA!")
print(f"Total de registros agregados: {len(full):,}")
print(f"Registros de novembro: {len(nov):,}")

display(full.head())
```

2. Tarifas históricas
Tarifa média ponderada por assentos e variação percentual em relação a novembro/2019.

```python
import matplotlib.pyplot as plt

def media_ponderada(df, coluna):
    return np.average(df[coluna].to_numpy(), weights=df.assentos.to_numpy())

resumo = nov.groupby('ano').apply(lambda g: pd.Series({
    'tarifa_media_real': media_ponderada(g, 'tarifa_media_real'),
    'tarifa_media_nominal': media_ponderada(g, 'tarifa_media_nominal'),
    'assentos': int(g.assentos.sum()),
    'combinacoes': len(g)
}), include_groups=False).reset_index()
base_2019 = float(resumo.loc[resumo.ano.eq(2019), 'tarifa_media_real'].iloc[0])
resumo['variacao_vs_2019_%'] = (resumo.tarifa_media_real/base_2019-1)*100
display(resumo.round(2))

fig, axs = plt.subplots(1,2,figsize=(12,4))
axs[0].plot(resumo.ano.astype(str),resumo.tarifa_media_real,marker='o')
axs[0].set_title('Tarifa média real — novembro')
axs[0].set_ylabel('R$ de dezembro/2025')
axs[0].grid(axis='y',alpha=.2)
axs[1].bar(resumo.ano.astype(str),resumo['variacao_vs_2019_%'])
axs[1].axhline(0,color='gray',linewidth=.8)
axs[1].set_ylabel('Variação real (%)')
axs[1].set_title('Variação frente a novembro/2019')
fig.tight_layout(); plt.show()
```

3. Treinar e comparar os seis modelos
Treino 2023: novembros de 2019, 2020 e 2022. Treino 2025: incluir novembro/2023. Métricas ponderadas por assentos. Menor MAE indica menor erro.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.dummy import DummyRegressor
from joblib import dump

ROTA = ['empresa','origem','destino']
COM_ANO = ['ano']+ROTA

def pipeline_modelo(algoritmo, com_ano=True):
    transformacoes = [('categoricas', OneHotEncoder(handle_unknown='ignore'), ROTA)]
    if com_ano: transformacoes.append(('ano', StandardScaler(), ['ano']))
    prep = ColumnTransformer(transformacoes)
    if algoritmo=='Ridge': modelo=Ridge(alpha=100)
    elif algoritmo=='Random Forest':
        modelo=RandomForestRegressor(n_estimators=70, max_depth=16,
                   min_samples_leaf=4, max_features=0.8, n_jobs=-1, random_state=42)
    else: raise ValueError('Algoritmo não reconhecido')
    return Pipeline([('prep',prep),('model',modelo)])

def avaliar_predicao(teste, predito):
    real=teste.tarifa_media_real.to_numpy()
    peso=teste.assentos.to_numpy()
    return {'MAE_R$':mean_absolute_error(real,predito,sample_weight=peso),
       'RMSE_R$':np.sqrt(mean_squared_error(real,predito,sample_weight=peso)),
       'R2':r2_score(real,predito,sample_weight=peso)}

def baseline_persistencia(treino, teste):
    ultimo=treino.loc[treino.ano.eq(int(treino.ano.max()))].copy()
    por_rota=ultimo.set_index(ROTA).tarifa_media_real
    por_empresa=ultimo.groupby('empresa').apply(
       lambda g: media_ponderada(g,'tarifa_media_real'),include_groups=False)
    geral=media_ponderada(ultimo,'tarifa_media_real')
    prev=[]; origens=[]
    for emp,orig,dest in teste[ROTA].itertuples(index=False,name=None):
        if (emp,orig,dest) in por_rota.index:
            prev.append(float(por_rota.loc[(emp,orig,dest)]));origens.append('Rota existente')
        elif emp in por_empresa.index:
            prev.append(float(por_empresa.loc[emp]));origens.append('Média da companhia')
        else:
            prev.append(float(geral));origens.append('Média global')
    return np.asarray(prev),pd.Series(origens,name='referencia')

def avaliar_periodo(ano,treino,teste):
    saida=[]; modelos={}; previsoes={}
    baseline=DummyRegressor(strategy='mean')
    baseline.fit(treino[COM_ANO],treino.tarifa_media_real,sample_weight=treino.assentos)
    p=baseline.predict(teste[COM_ANO]);modelos['Média histórica']=baseline
    previsoes['Média histórica']=p
    saida.append({'ano':ano,'modelo':'Média histórica',**avaliar_predicao(teste,p)})
    p, fontes=baseline_persistencia(treino,teste)
    previsoes['Persistência por rota']=p
    saida.append({'ano':ano,'modelo':'Persistência por rota',**avaliar_predicao(teste,p)})
    print(f'{ano}: cobertura persistência:')
    display((fontes.value_counts(normalize=True)*100).round(2).rename('% grupos'))
    for alg in ['Ridge','Random Forest']:
        for com_ano in [True,False]:
            nome=f'{alg} ({"com ano" if com_ano else "sem ano"})'
            atributos=COM_ANO if com_ano else ROTA
            modelo=pipeline_modelo(alg,com_ano)
            modelo.fit(treino[atributos],treino.tarifa_media_real,
                       model__sample_weight=treino.assentos)
            p=modelo.predict(teste[atributos])
            modelos[nome]=modelo; previsoes[nome]=p
            saida.append({'ano':ano,'modelo':nome,**avaliar_predicao(teste,p)})
    return pd.DataFrame(saida),modelos,previsoes

TREINOS={2023:nov[nov.ano.isin([2019,2020,2022])],
        2025:nov[nov.ano.isin([2019,2020,2022,2023])]}
TESTES={2023:nov[nov.ano.eq(2023)],2025:nov[nov.ano.eq(2025)]}
metricas_periodos=[]; modelos_treinados={}; previsoes_periodos={}
for ano in [2023,2025]:
    m,mods,preds=avaliar_periodo(ano,TREINOS[ano],TESTES[ano])
    metricas_periodos.append(m)
    modelos_treinados[ano]=mods
    previsoes_periodos[ano]=preds
metricas=pd.concat(metricas_periodos,ignore_index=True)
display(metricas.round(3))
metricas.to_csv('/content/metricas_modelos_consolidadas.csv',index=False)
print('Melhor MAE em cada período:')
display(metricas.loc[metricas.groupby('ano')['MAE_R$'].idxmin(),['ano','modelo','MAE_R$','R2']].round(3))
```

4. Analisar generalização
Diagnóstico de combinações novas de companhia–origem–destino e comparação visual dos erros dos modelos.

```python
diagnosticos=[]
for ano in [2023,2025]:
    treino,teste=TREINOS[ano],TESTES[ano]
    vistos=set(map(tuple,treino[ROTA].to_numpy()))
    inedito=~teste[ROTA].apply(tuple,axis=1).isin(vistos)
    pred=previsoes_periodos[ano]['Random Forest (com ano)']
    for grupo,mascara in [('Já observada',~inedito.to_numpy()),('Inédita',inedito.to_numpy())]:
        sub=teste.loc[mascara]
        if len(sub):
            diagnosticos.append({'ano':ano,'grupo':grupo,'combinacoes':len(sub),
               'assentos':int(sub.assentos.sum()),
               'MAE_RF_com_ano_R$':avaliar_predicao(sub,np.asarray(pred)[mascara])['MAE_R$']})
diagnostico=pd.DataFrame(diagnosticos)
diagnostico['pct_assentos_ano']=diagnostico.assentos/diagnostico.groupby('ano').assentos.transform('sum')*100
display(diagnostico.round(3))
diagnostico.to_csv('/content/diagnostico_combinacoes.csv',index=False)

fig,axs=plt.subplots(1,2,figsize=(13,5))
for ax,ano in zip(axs,[2023,2025]):
    dados=metricas[metricas.ano.eq(ano)].sort_values('MAE_R$')
    ax.barh(dados.modelo,dados['MAE_R$'])
    ax.invert_yaxis();ax.set_title(f'Novembro/{ano} — MAE ponderado')
    ax.set_xlabel('R$ (menor é melhor)')
fig.tight_layout();plt.show()
```

5. Painel interativo para o analista
Um painel compacto com histórico e rotas, desempenho dos modelos e simulação experimental. Estimativas não são cotações.

```python
import gradio as gr

empresas = sorted(nov.empresa.unique().tolist())
origens = sorted(nov.origem.unique().tolist())
destinos = sorted(nov.destino.unique().tolist())

def consulta(empresa, origem, destino):
    df = nov.copy()
    for coluna, valor in [('empresa',empresa),('origem',origem),('destino',destino)]:
        if valor != 'Todos': df = df[df[coluna].eq(valor)]
    if df.empty: return pd.DataFrame(), None
    tab = df.groupby('ano').apply(lambda g: pd.Series({
        'tarifa_real_R$': media_ponderada(g,'tarifa_media_real'),
        'assentos':int(g.assentos.sum()), 'grupos':len(g)
    }), include_groups=False).reset_index()
    fig, ax = plt.subplots(figsize=(7,3.3))
    ax.plot(tab.ano.astype(str), tab['tarifa_real_R$'], marker='o')
    ax.set_title('Tarifa média real — novembro');ax.set_ylabel('R$ dez/2025')
    ax.grid(axis='y', alpha=.25);fig.tight_layout()
    return tab.round(2),fig

def resultado(ano):
    tab=metricas[metricas.ano.eq(int(ano))].sort_values('MAE_R$')
    return tab.round(3), diagnostico[diagnostico.ano.eq(int(ano))].round(3)

def simular(empresa,origem,destino):
    if origem==destino:return pd.DataFrame(),pd.DataFrame(), 'Selecione aeroportos diferentes.'
    entrada=pd.DataFrame([{'ano':2025,'empresa':empresa,'origem':origem,'destino':destino}])
    nomes=['Média histórica','Ridge (com ano)','Ridge (sem ano)',
           'Random Forest (com ano)','Random Forest (sem ano)']
    previsoes=[]
    for nome in nomes:
        col=ROTA if 'sem ano' in nome else COM_ANO
        val=float(modelos_treinados[2025][nome].predict(entrada[col])[0])
        previsoes.append({'modelo':nome,'estimativa_R$':round(val,2)})
    historico=nov[(nov.empresa==empresa)&(nov.origem==origem)&(nov.destino==destino)]
    historico=historico[['ano','tarifa_media_real','assentos']].sort_values('ano')
    return pd.DataFrame(previsoes), historico, 'Estimativas de novembro/2025 em R$ dez/2025; não são cotações.'

with gr.Blocks(title='SSD ANAC — Painel do analista') as painel:
    gr.Markdown('# SSD ANAC — Análise de tarifas\n**Valores históricos ANAC/IPCA; não são ofertas comerciais.**')
    with gr.Tab('Histórico'):
        with gr.Row():
            e=gr.Dropdown(['Todos']+empresas,value='Todos',label='Companhia')
            o=gr.Dropdown(['Todos']+origens,value='Todos',label='Origem')
            d=gr.Dropdown(['Todos']+destinos,value='Todos',label='Destino')
        b=gr.Button('Analisar'); t=gr.Dataframe(label='Tarifas por novembro'); f=gr.Plot()
        b.click(consulta,[e,o,d],[t,f])
    with gr.Tab('Modelos e erros'):
        a=gr.Radio([2023,2025],value=2025,label='Ano de teste')
        b2=gr.Button('Comparar'); m=gr.Dataframe(label='Métricas ponderadas')
        g=gr.Dataframe(label='Combinações inéditas')
        b2.click(resultado,[a],[m,g])
    with gr.Tab('Simulador experimental'):
        with gr.Row():
            se=gr.Dropdown(empresas,value=empresas[0],label='Companhia')
            so=gr.Dropdown(origens,value=origens[0],label='Origem')
            sd=gr.Dropdown(destinos,value=destinos[-1],label='Destino')
        bs=gr.Button('Simular');sp=gr.Dataframe(label='Estimativas dos modelos')
        sh=gr.Dataframe(label='Histórico desta combinação');msg=gr.Textbox(label='Observação')
        bs.click(simular,[se,so,sd],[sp,sh,msg])

painel.launch(share=True)
```
